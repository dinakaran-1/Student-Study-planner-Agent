# Study Planner Agent Demo

**Purpose:** prove that this project is an agent, not a chatbot.

Each example shows the final answer plus the multi-step tool trace. The agent decides which tools to call, receives their results, and decides what to do next. The project also stores plans and preferences in persistent memory.

**Before running:** install `requirements.txt` and create a local `.env` containing `GEMINI_API_KEY`. Never place the real API key in the notebook or submission ZIP.

```python
from agent import StudyPlannerAgent

agent = StudyPlannerAgent()
print("Agent initialized successfully.")
```

```python
def demo(goal):
    print("\n" + "=" * 80)
    print("GOAL:", goal)
    print("=" * 80)
    result = agent.run(goal)
    print("\nFINAL ANSWER:\n")
    print(result["answer"])
    print("\nMULTI-STEP TOOL TRACE:\n")
    for step in result["tool_trace"]:
        print(f"Step {step['step']}: {step['tool']}")
        print("Arguments:", step["arguments"])
        print("Result:", step["result"])
        print("-" * 60)
    return result
```

## Example 1: Multi-subject exam plan

```python
result1 = demo(
    "Create a 14-day study plan for Python, SQL and aptitude. "
    "I can study 3 hours a day and Python is my weakest subject."
)
```

## Example 2: Practical preparation

```python
result2 = demo(
    "Create a 7-day intensive plan for my Python practical. "
    "I have 3 hours each day and need more hands-on coding practice."
)
```

## Example 3: Different subject domain

```python
result3 = demo(
    "Create a 10-day study plan for Physics, Mathematics and English. "
    "I have 2 hours per day."
)
```

## Why this is an agent

The traces above show multiple tool calls and steps. The model decides which tool to call, Python executes it, the tool result is returned to the model, and the model decides what to do next. Saved plans and preferences are persisted in `study_memory.json`, providing memory across turns and app restarts.

## Submission evidence

- **Agent:** the model selects tools instead of only generating text.
- **Multi-step loop:** each demo prints the ordered tool calls and results.
- **Tools:** `analyze_subjects()` and `build_schedule()` are the two main planning tools.
- **Memory:** saved plans and preferences persist in `study_memory.json`.
- **Generality:** examples cover programming and non-programming subjects.
